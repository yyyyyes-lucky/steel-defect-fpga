```python
from pathlib import Path
import sys
root = Path.cwd().resolve()
if not (root / 'steel_static_demo_20261029').is_dir():
    root = root / 'steel_board_runtime'
package = root / 'steel_dynamic_demo_20261028'
if not package.is_dir():
    raise FileNotFoundError('Open this notebook inside the extracted steel_board_runtime folder: ' + str(root))
sys.path.insert(0, str(package))
import json
settings_path = root / 'dynamic_settings.json'
settings = json.loads(settings_path.read_bytes()) if settings_path.is_file() else {}
if not isinstance(settings, dict) or set(settings) - {'port', 'max_frames', 'idle_seconds'}:
    raise ValueError('dynamic_settings.json only accepts port, max_frames, idle_seconds')
options = dict(port=8767, max_frames=1000, idle_seconds=600)
options.update(settings)
if type(options['port']) is not int or not 1024 <= options['port'] <= 65535:
    raise ValueError('port must be an integer between 1024 and 65535')
if type(options['max_frames']) is not int or not 1 <= options['max_frames'] <= 10000:
    raise ValueError('max_frames must be an integer between 1 and 10000')
if type(options['idle_seconds']) not in (int, float) or not 30 <= options['idle_seconds'] <= 3600:
    raise ValueError('idle_seconds must be between 30 and 3600')
print('Board API only. On Windows run start_dynamic.cmd, then open the printed DYNAMIC_WEB address.')
from board_service import serve
report = serve(**options)
```
